# 一、Docker 镜像相关命令
1、查看镜像: docker images (-q 只查看ID)  
IMAGE      ID             DISK USAGE   CONTENT SIZE   EXTRA  
centos:7   be65f488b776        301MB         76.1MB    U  
2、搜索镜像: docker search + 镜像名称(redis)  
3、拉取镜像: docker pull redis:3.2(hub.docker.com去找)  
4、删除镜像: docker rmi + ID(或redis:3.2)  
5、不同镜像:  
centos:7——基础系统镜像  
redis:3.2——Redis 服务镜像  
mysql:5.7——MySQL 数据库服务镜像  
nginx:latest——Nginx Web 服务器镜像

# 二、Docker 容器相关命令
1、查看容器: docker ps（正在运行的容器） 加一个-a 查历史
2、创建容器: docker run -it --name=c1 centos:7 /bin/bash（进入容器内部，使用exit可以退出。如果使用-id，则后台也会运行）  
<img src="../pics/docker_run.png" width="65%" />  
3、进入容器: docker exec -it +name /bin/bash（先启动才能进）
4、启动容器: docker start +name
5、停止容器: docker stop +name
6、删除容器: docker rm +name/id
7、查看容器信息: docker inspcet +name

# 三、docker 容器的数据卷
1、基本概念：数据卷是宿主机中的一个目录或文件，当容器目录和数据卷目录绑定后，对方的修改会立即同步（一个数据卷可以被多个容器同时挂载）  
2、配置数据卷: docker run ... -v +宿主机目录:容器内目录（必须是绝对路径，不存在则会自动创建，可以多个-v挂载多个数据卷）
